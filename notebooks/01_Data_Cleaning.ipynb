# 01 - Data Cleaning

The purpose of this notebook is to inspect and clean the used car dataset before performing exploratory data analysis and building machine learning models.

The dataset contains information about used cars, including their name, manufacturing year, selling price, kilometres driven, fuel type, seller type, transmission, and ownership history.

The main objectives are:
- Understand the structure of the dataset.
- Identify missing values and duplicate records.
- Remove unnecessary duplicate records.
- Create a new feature representing the age of each car.
- Save the cleaned dataset for further analysis.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from pathlib import Path

file_path = Path("..") / "data" /"CAR DETAILS FROM CAR DEKHO.csv"
df = pd.read_csv(file_path)
display(df.head())
print(f"Shape: {df.shape}")
```

## 1. Initial Data Inspection

Before cleaning the dataset, we examine its quality using the following checks:

- **Data Types:** Identify the data type of each column.
- **Missing Values:** Check whether any columns contain null values.
- **Unique Values:** Understand the number of distinct values in each column.
- **Duplicate Records:** Identify repeated rows in the dataset.

These checks help determine which cleaning operations are necessary.


```python
quality_summary = pd.DataFrame({
    "Data Type": df.dtypes.astype(str),
    "Missing Values": df.isna().sum(),
    "Unique Values": df.nunique()
})

display(quality_summary)

print(f"Total null values: {df.isnull().sum().sum()}")

print(f"\nDuplicate rows: {df.duplicated().sum()}")
```

```python
df = df.dropna().drop_duplicates().reset_index(drop=True)
print(f"\nShape after cleaning: {df.shape}")
```


Since there were no missing values, removing null-containing rows did not affect the dataset.
Duplicate records were removed using `drop_duplicates()`. The index was then reset to maintain a continuous sequence.
This reduced the dataset from 4,340 records to 3,577 records.

## 2. Feature Creation - Car Age

A new feature named `car_age` was created using the manufacturing year.
The calculation is:
`car_age = 2026 - year`

This represents the approximate age of each vehicle in years, using 2026 as the reference year.
Creating this feature provides another way to analyze the relationship between vehicle age and selling price.

```python
df["car_age"] = 2026 - df["year"]
df.head(5)
```

```python
output_path = Path("..") / "Data" / "car_data_cleaned.csv"

output_path.parent.mkdir(parents=True, exist_ok=True)

df.to_csv(output_path, index=False)

print("Cleaned dataset saved")
```

## 3. Conclusion

The dataset was inspected, duplicate records were removed, and a new feature called `car_age` was created.
The final dataset contains 3,577 records and 9 columns, compared with the original 4,340 records and 8 columns.
This cleaned dataset will serve as the input for the next stage: Exploratory Data Analysis (EDA).
